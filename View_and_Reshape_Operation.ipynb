```python
import torch
```

```python
tensor= torch.arange(12)
```

```python
reshaped_tensor_view=tensor.view(2,6)
reshaped_tensor_reshape=tensor.reshape(2,6)
print(reshaped_tensor_view)
print(reshaped_tensor_reshape)
```

```python
reshaped_tensor_view=tensor.view(2,6)
reshaped_tensor_reshape=tensor.reshape(2,6)
print(reshaped_tensor_view)
print(reshaped_tensor_reshape)
print(reshaped_tensor_view.shape)
print(reshaped_tensor_reshape.shape)
```

```python
tensor.is_contiguous()
```

```python
inferred_shape= tensor.view(3,-1)
print(inferred_shape)
print(inferred_shape.shape)
```

```python
tensor3d= torch.arange(24).reshape(2,3,4)
print(tensor3d)
print(tensor3d.shape)
```

```python
flattened= tensor3d.view(-1)
print(flattened)
```

```python
flattened.is_contiguous()
```

```python
tensor2d= torch.arange(24).reshape(12,2)
transpose_tensor= tensor2d.t()
```

```python
transpose_tensor.is_contiguous()
```

```python
error = transpose_tensor.view(6,4)
```

```python
error= transpose_tensor.reshape(6,4)
```

```python
print(error)
print(error.shape)
```

```python
contiguous= transpose_tensor.contiguous()
error= contiguous.view(6,4)
```

```python
print(error)
print(error.shape)
```
